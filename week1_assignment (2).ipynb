```python
########### YOUR CODE GOES HERE ###########
########### NORMALIZE ALL ROWS ###########

import re
from datetime import date

MONTH = (
    r"(?:January|February|March|Marc|April|May|June|July|"
    r"August|September|Sept|October|November|December|"
    r"Jan|Feb|Mar|Apr|Jun|Jul|Aug|Sep|Oct|Nov|Dec)"
)

patterns = [
    # 7/21/1999 or 7-21-99
    rf"(?<!\d)\d{{1,2}}[/-]\d{{1,2}}[/-]\d{{2,4}}(?!\d)",

    # September 27, 2021
    rf"{MONTH}\.?\s+\d{{1,2}}(?:st|nd|rd|th)?[,]?\s+\d{{4}}",

    # 26 May 2010
    rf"(?<!\d)\d{{1,2}}\s+{MONTH}\.?,?\s+\d{{4}}",

    # 09/1975 or 9-2009
    rf"(?<!\d)\d{{1,2}}[/-]\d{{4}}(?!\d)",

    # March 1979 or Sep 2007
    rf"{MONTH}\.?,?\s+\d{{4}}(?!\d)",

    # 2010
    r"(?<!\d)(?:19|20)\d{2}(?!\d)"
]

month_names = {
    "jan": 1, "feb": 2, "mar": 3,
    "apr": 4, "may": 5, "jun": 6,
    "jul": 7, "aug": 8, "sep": 9,
    "oct": 10, "nov": 11, "dec": 12
}

def normalize_date(raw_date):
    s = str(raw_date).strip().replace(".", "")
    s = re.sub(r"(\d)(st|nd|rd|th)\b", r"\1", s)
    s = re.sub(r",", " ", s)
    s = re.sub(r"\s+", " ", s)

    # Numeric dates
    if "/" in s or "-" in s:
        parts = re.split(r"[/-]", s)
        month = int(parts[0])

        if len(parts) == 3:
            day = int(parts[1])
            year = int(parts[2])
        else:
            day = 1
            year = int(parts[1])

    else:
        parts = s.split()

        if len(parts) == 1:
            return date(int(parts[0]), 1, 1).isoformat()

        if parts[0].isdigit():
            day = int(parts[0])
            month = month_names[parts[1][:3].lower()]
            year = int(parts[2])
        else:
            month = month_names[parts[0][:3].lower()]

            if len(parts) == 3:
                day = int(parts[1])
                year = int(parts[2])
            else:
                day = 1
                year = int(parts[1])

    if year < 100:
        year += 1900

    return date(year, month, day).isoformat()


# Rebuild all rows from original text
extracted_contents = {}

for line in lines:
    if not line.strip():
        continue

    row_index, content = line.split("\t", 1)
    row_index = int(row_index)

    candidates = []

    for priority, pattern in enumerate(patterns):
        for match in re.finditer(pattern, content, re.I):
            try:
                normalized = normalize_date(match.group())
                candidates.append(
                    (match.start(), priority, normalized)
                )
            except (ValueError, KeyError, IndexError):
                continue

    if candidates:
        extracted_contents[row_index] = min(candidates)[2]


# Normalize and validate
output = []

for row_index in sorted(extracted_contents):
    normalized = extracted_contents[row_index]
    output.append(f"{row_index}\t{normalized}")

print("Dates extracted:", len(output))

missing = sorted(set(range(500)) - set(extracted_contents))
print("Missing row numbers:", missing)

assert len(output) == 500, (
    f"Expected 500 rows, received {len(output)}"
)

assert sorted(extracted_contents) == list(range(500))

assert all(
    re.fullmatch(r"\d+\t\d{4}-\d{2}-\d{2}", row)
    for row in output
)

print("All 500 rows passed structural validation.")
print(*output[:10], sep="\n")

```
